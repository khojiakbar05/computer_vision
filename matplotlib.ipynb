```python
import matplotlib.pyplot as plt
# print(matplotlib.__version__)
import numpy as np
```

# Basic Line graph

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 25, 30])

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

# Line Styles and colors

```python
x = np.array([1, 2, 3, 4, 6])
y = np.array([3, 9, 2, 13, 7])

plt.plot(y, linestyle="dotted", color="green", marker="o")
plt.title("Styled Line Example")
plt.show()
```

# Multiple Lines in One Graph with Numpy and Mapplotib


```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([1, 3, 5, 7])

plt.plot(x, y1, label="Line 1", color="red", marker="s")
plt.plot(x, y2, label="Line 2", color="blue", linestyle="--", marker="o")
plt.title("Multiple Lines Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.legend() # line markerini va styleni korsatib turadi
plt.show()
```

# Bar Chart with Numpy and Maplotlib

```python
x = np.array(['Salary', 'Bill', 'Charity', 'Donation'])
y = np.array([5, 9, 7, 1])

plt.bar(x, y, color="orange")
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

# Scatter Plot with Numoy and Matplotlib

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 2, 4, 2, 1])

plt.scatter(x, y, color="purple", s=100)
plt.title("Scatter Plot Example")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

# Histogram with Numpy and Matplotlib

```python
data = np.array([1, 2, 2, 3, 3, 3, 4, 4, 5])

plt.hist(data, bins=5, color="skyblue", edgecolor="black")
plt.title("Histogram Example")
plt.xlabel("Values")
plt.ylabel("Count")
plt.show()
```

# Subplots(Multiple Graphs Together)

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([1, 3, 5, 7])

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color="red")
plt.title("Line 1")

plt.subplot(1, 2, 2)
plt.bar(x, y2, color="blue")
plt.title("Line 2")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data for a double helix
n = 50
theta = np.linspace(0, 2*np.pi, n)
x1 = np.cos(theta)
y1 = np.sin(theta)
z1 = np.linspace(0, 1, n)
x2 = np.cos(theta + np.pi)
y2 = np.sin(theta + np.pi)
z2 = z1

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.fill_between(x1, y1, z1, x2, y2, z2, alpha=0.5)
ax.plot(x1, y1, z1, linewidth=2, color='C0')
ax.plot(x2, y2, z2, linewidth=2, color='C0')

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
np.random.seed(19680801)
n = 100
rng = np.random.default_rng()
xs = rng.uniform(23, 32, n)
ys = rng.uniform(0, 100, n)
zs = rng.uniform(-50, -25, n)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.scatter(xs, ys, zs)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
